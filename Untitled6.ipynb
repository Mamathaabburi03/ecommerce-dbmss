```python
import sqlite3
```

```python
conn = sqlite3.connect("ecommerce.db")

cursor = conn.cursor()
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS users (
    user_id INTEGER PRIMARY KEY,
    name TEXT,
    email TEXT
)
""")

conn.commit()
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS products (
    product_id INTEGER PRIMARY KEY,
    product_name TEXT,
    price INTEGER
)
""")

conn.commit()
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS orders (
    order_id INTEGER PRIMARY KEY,
    user_id INTEGER,
    product_id INTEGER,
    quantity INTEGER
)
""")

conn.commit()
```

```python
users = [
    ('Mamatha', 'mamatha@gmail.com'),
    ('sam', 'sam@gmail.com')
]

cursor.executemany("""
INSERT INTO users (name, email)
VALUES (?, ?)
""", users)

conn.commit()
```

```python
products = [
    ('Laptop', 60000),
    ('Mobile', 25000),
    ('Headphones', 3000)
]

cursor.executemany("""
INSERT INTO products (product_name, price)
VALUES (?, ?)
""", products)

conn.commit()
```

```python
orders = [
    (1, 1, 1),
    (2, 2, 2)
]

cursor.executemany("""
INSERT INTO orders (user_id, product_id, quantity)
VALUES (?, ?, ?)
""", orders)

conn.commit()
```

```python
cursor.execute("""
SELECT users.name,
products.product_name,
orders.quantity

FROM orders

JOIN users
ON orders.user_id = users.user_id

JOIN products
ON orders.product_id = products.product_id
""")

rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
cursor.execute("""
SELECT SUM(products.price * orders.quantity)
FROM orders

JOIN products
ON orders.product_id = products.product_id
""")

sales = cursor.fetchone()

print("Total Sales:", sales[0])
```

```python
conn.close()
```
